# Agent Supervisor

This notebook walks through the `agent-supervisor` example: an LLM-driven
supervisor that routes between a `researcher` agent (web search) and a
`coder` agent (python execution / charting).

It fleshes out the `agent_supervisor.py` template in `examples/templates/`,
and mirrors LangGraph's own "agent supervisor" pattern. Compare this to the
`multi-agent-collaboration` example, which uses fixed transitions instead of
a supervisor deciding routing at each step.

## Build the application

We reuse the `main()` function from `application.py`, which wires up the actions and transitions.

```python
import application

app = application.main(
    "Fetch the UK's GDP over the past 5 years, then draw a line graph of it."
)
```

## Visualize the state machine

This shows the supervisor's routing to `researcher`/`coder`, and their shared `tool_node`.

```python
app.visualize(
    output_file_path="statemachine", include_conditions=True, view=True, format="png"
)
```

## Run the application

This runs until the `terminal` action is reached (i.e. the supervisor decides `FINISH`).

```python
action, result, state = app.run(halt_after=["terminal"])
```

## Inspect the conversation

Each message shows which agent said what, and any tool calls/results along the way.

```python
for message in state["messages"]:
    print(message)
```
