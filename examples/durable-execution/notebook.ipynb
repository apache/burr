```python
# Build the durable-execution application and run until it suspends.
import sys, pathlib
sys.path.insert(0, str(pathlib.Path(".").resolve()))

from application import build_application
from burr.core import resume

app, graph, persister = build_application(app_id="notebook-demo-1", db_path=":memory:")
app.run(halt_after=["review"])
print("Suspended channel:", app.suspended.channel)
print("Metadata:", app.suspended.metadata)
```

```python
# Simulate the human approving and resume the workflow.
final_state = resume(
    persister=persister,
    graph=graph,
    app_id="notebook-demo-1",
    partition_key=None,
    channel="human_approval",
    payload={"approved": True},
)
print("Approved:", final_state["approved"])
```
